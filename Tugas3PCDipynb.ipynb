```python
# IJAZAH 1

import cv2
import numpy as np
import matplotlib.pyplot as plt

# 1. Baca citra asli (default OpenCV adalah BGR, ubah ke RGB)
img_bgr = cv2.imread('/content/sample_data/foto.jpg')
img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)

# 2. Konversi ke Grayscale
img_gray = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2GRAY)

# 3. Tampilkan gambar dan histogram
plt.figure(figsize=(12, 8))

# Tampilkan Citra RGB & Grayscale
plt.subplot(2, 2, 1)
plt.imshow(img_rgb)
plt.title('Citra Asli (RGB)')
plt.axis('off')

plt.subplot(2, 2, 2)
plt.imshow(img_gray, cmap='gray')
plt.title('Citra Grayscale')
plt.axis('off')

# Plot Histogram RGB
plt.subplot(2, 2, 3)
warna = ('r', 'g', 'b')
for i, col in enumerate(warna):
    hist_color = cv2.calcHist([img_rgb], [i], None, [256], [0, 256])
    plt.plot(hist_color, color=col)
plt.title('Histogram RGB')
plt.xlim([0, 256])

# Plot Histogram Grayscale
plt.subplot(2, 2, 4)
hist_gray = cv2.calcHist([img_gray], [0], None, [256], [0, 256])
plt.plot(hist_gray, color='black')
plt.title('Histogram Grayscale')
plt.xlim([0, 256])

plt.tight_layout()
plt.show()


# 4. Enhancement dengan Contrast Stretching
min_val = np.min(img_gray)
max_val = np.max(img_gray)

img_enhanced = ((img_gray - min_val) / (max_val - min_val) * 255).astype(np.uint8)


# 5. Tampilkan hasil enhancement
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(img_gray, cmap='gray')
plt.title('Sebelum Enhancement')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(img_enhanced, cmap='gray')
plt.title('Setelah Contrast Stretching')
plt.axis('off')

plt.tight_layout()
plt.show()


# 6. Histogram setelah enhancement
plt.figure(figsize=(8, 5))

hist_enhanced = cv2.calcHist(
    [img_enhanced],
    [0],
    None,
    [256],
    [0, 256]
)

plt.plot(hist_enhanced, color='black')

plt.title('Histogram Setelah Contrast Stretching')
plt.xlabel('Intensitas Piksel')
plt.ylabel('Frekuensi')
plt.xlim([0, 256])

plt.show()
```
